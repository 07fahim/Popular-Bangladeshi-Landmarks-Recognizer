```python
import random
import shutil
from pathlib import Path


data_path = Path(r"F:\dataset_landmarks")

work_path = Path(r"F:\landmarks_working")

# Test folder inside the working directory
test_path = work_path / "test"

if not work_path.exists():
    shutil.copytree(data_path, work_path)  
    print(f"Copied dataset to {work_path}")

# Create test directory
test_path.mkdir(exist_ok=True)

test_split = 0.1  # 10% of images to test
random.seed(42)

for class_dir in work_path.iterdir():
    if class_dir.is_dir() and class_dir.name.lower() != "test":
        images = list(class_dir.glob("*"))
        n_test = int(len(images) * test_split)
        if n_test == 0:
            continue  

        test_images = random.sample(images, n_test)
        class_test_dir = test_path / class_dir.name
        class_test_dir.mkdir(parents=True, exist_ok=True)

        for img in test_images:
            shutil.move(str(img), str(class_test_dir / img.name))

        print(f"Moved {n_test} images from '{class_dir.name}' → 'test/{class_dir.name}'")

print("✅ Test split completed!")
```
